```python
import pandas as pd
import re

# Load Romanized Sinhala datasets
train_df = pd.read_csv('SOLD_dataset/SOLD_train.tsv', sep='\t', encoding='utf-8')
test_df = pd.read_csv('SOLD_dataset/SOLD_test.tsv', sep='\t', encoding='utf-8')

# Combine them
full_df = pd.concat([train_df, test_df], ignore_index=True)

romanized_tweets = full_df['text'].dropna().tolist()

# Token stats
total_words = 0
total_english = 0
total_sinhala = 0

def is_english_word(word):
    return re.fullmatch(r'[A-Za-z]+', word) is not None

for tweet in romanized_tweets:
    words = tweet.split()
    total_words += len(words)
    english_count = sum(1 for w in words if is_english_word(w))
    sinhala_count = len(words) - english_count
    total_english += english_count
    total_sinhala += sinhala_count

tweet_count = len(romanized_tweets)

avg_words_per_tweet = total_words / tweet_count
avg_english_tokens = total_english / tweet_count
avg_sinhala_tokens = total_sinhala / tweet_count
code_mix_ratio = (
    (avg_sinhala_tokens / avg_words_per_tweet) * 100,
    (avg_english_tokens / avg_words_per_tweet) * 100
)


# Results
print(f"Total tweets: {tweet_count}")
print(f"Total words in dataset: {total_words}")
print(f"Average words per tweet: {avg_words_per_tweet:.2f}")
print(f"Average Romanized Sinhala tokens per tweet: {avg_sinhala_tokens:.2f}")
print(f"Average English tokens per tweet: {avg_english_tokens:.2f}")
print(f"Code-mixed ratio (Romanized : English): {code_mix_ratio[0]:.1f}% : {code_mix_ratio[1]:.1f}%")
```
